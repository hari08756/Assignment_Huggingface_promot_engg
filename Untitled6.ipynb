```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

print("CUDA available:", torch.cuda.is_available())
```

```python
MODEL_NAME = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

chat_pipeline = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=300,
    do_sample=True,
    temperature=0.3,
    top_p=0.9,
    repetition_penalty=1.1,
)
print("Model loaded:", MODEL_NAME)
```

```python
def ask_assistant(user_message: str, system_prompt: str = None) -> str:
    """Send a message to the chatbot using the model's chat template."""
    if system_prompt is None:
        system_prompt = (
            "You are a cautious clinical assistant. You provide general medical "
            "information only, you never give a definitive diagnosis, and you "
            "always recommend consulting a licensed physician for confirmation."
        )
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message},
    ]
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    output = chat_pipeline(prompt)[0]["generated_text"]
    # strip the prompt portion so only the new completion is returned
    return output[len(prompt):].strip()

# Example interactive-style call
response = ask_assistant("Patient has fever for 3 days.")
print("User: Patient has fever for 3 days.\n")
print("Assistant:", response)
```

```python
import json

task5_prompt = """
Return ONLY a valid JSON object, with no extra text, no markdown fences, and no
explanation before or after it. Use exactly this schema and key order:

{
  "disease": "",
  "symptoms": [],
  "risk_level": "",
  "recommendation": ""
}

Patient data:
- Fever: present
- Cough: present
- Oxygen Level: 88%

If you are not certain of the exact disease, use your best clinical judgement
and state that uncertainty inside the "recommendation" field rather than
inventing a false certainty.
"""

response_task5 = ask_assistant(task5_prompt)
print("Raw model output:\n", response_task5)

# Defensive parsing since small open models may not perfectly follow format
try:
    parsed = json.loads(response_task5)
    print("\nParsed JSON:\n", json.dumps(parsed, indent=2))
except json.JSONDecodeError:
    print("\n(Model did not return perfectly valid JSON — this is common with")
    print("small models; a larger instruct model or a JSON-mode API call")
    print("would improve reliability. Consider adding a regex/json-repair step.)")
```

```python
ROLE_SYSTEM_PROMPTS = {
    "doctor": (
        "You are a clinical decision-support assistant for an attending "
        "physician. Use precise medical terminology, mention relevant "
        "differentials, and be concise. Assume strong clinical background."
    ),
    "nurse": (
        "You are a clinical assistant supporting a nurse. Focus on "
        "observable signs, monitoring parameters, and care/escalation "
        "protocols. Use standard clinical terminology but keep it practical."
    ),
    "medical_student": (
        "You are a teaching assistant for a medical student. Explain the "
        "underlying reasoning and pathophysiology in a structured, "
        "educational way, as if preparing them for an exam."
    ),
    "patient": (
        "You are a friendly assistant explaining health information to a "
        "patient with no medical background. Avoid jargon, use plain "
        "language, be reassuring but honest, and always recommend they "
        "confirm anything important with their doctor."
    ),
}

def ask_by_role(user_message: str, role: str) -> str:
    role = role.lower().replace(" ", "_")
    if role not in ROLE_SYSTEM_PROMPTS:
        raise ValueError(f"Unknown role: {role}. Choose from {list(ROLE_SYSTEM_PROMPTS)}")
    return ask_assistant(user_message, system_prompt=ROLE_SYSTEM_PROMPTS[role])

# Example: same question, four different roles
question = "The patient's oxygen level is 88%. What does that mean?"
for role in ROLE_SYSTEM_PROMPTS:
    print(f"=== Role: {role} ===")
    print(ask_by_role(question, role))
    print()
```
